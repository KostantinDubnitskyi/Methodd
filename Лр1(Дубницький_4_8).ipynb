## Завдання 2. Скалярний добуток і довжина векторів

Дано вектори:

2. Знайдіть евклідову норму кожного вектора.
3. Обчисліть косинус кута між векторами.

````python
u = np.array([2, -3, 6])
v = np.array([5,  1, 4])

#1 Обчисліть скалярний добуток двома способами: np.dot() і оператором @.
dot_np = np.dot(u, v)
dot_sop = u @ v
print("dot_np =", dot_np)
print("dot_sop =", dot_sop)

#2 Знайдіть евклідову норму кожного вектора.
norma_u = np.linalg.norm(u)
norma_v = np.linalg.norm(v)
print("norma_u =", norma_u)
print("norma_v =", norma_v)

#3 Обчисліть косинус кута між вектора
cos_angle = dot_np / (norma_u * norma_v)
print("cos(angle) =", cos_angle)
````

## Завдання 4. Матричне множення

Дано матриці:
```python
A = np.array([
    [2, 1, 3],
    [0, 4, 5]
])

B = np.array([
    [1, 2],
    [3, 0],
    [4, 1]
])
```

1. Обчисліть матричний добуток `A @ B`.
2. Визначте форму результату.
3. Обчисліть елемент результату з індексами `[0, 1]` вручну через відповідний рядок `A` і стовпець `B` та перевірте його.

````python
A = np.array([
    [2, 1, 3],
    [0, 4, 5]
])
B = np.array([
    [1, 2],
    [3, 0],
    [4, 1]
])

#1 Обчисліть матричний добуток A @ B.
C = A @ B
print("A @ B =", C)

#2 Визначте форму результату.
print("Форма результата:", C.shape)

#3 Обчисліть елемент результату з індексами [0, 1] вручну через відповідний рядок A і стовпець B та перевірте його.
row0 = A[0, :]
col1 = B[:, 1]
map = np.dot(row0, col1)
print("Рядок 0 матриці A:", row0)
print("Стовбчик 1 матриці B:", col1)
print("Ручний розрахунок C[0,1] =", map)
print("Из відповіді, C[0,1] =", C[0, 1])
print("Співпадають:", map == C[0, 1])
````

## Завдання 11. Перестановка рядків і стовпців

Дано матрицю:
```python
A = np.array([
    [11, 12, 13, 14],
    [21, 22, 23, 24],
    [31, 32, 33, 34],
    [41, 42, 43, 44]
])

row_order = np.array([2, 0, 3, 1])
col_order = np.array([3, 1, 0, 2])
```

1. Переставте рядки відповідно до `row_order`.
2. Після цього переставте стовпці відповідно до `col_order`.
3. За допомогою обернених перестановок відновіть початкову матрицю.

Для побудови оберненої перестановки використайте `np.argsort()`.

````python
import numpy as np

A = np.array([
    [11, 12, 13, 14],
    [21, 22, 23, 24],
    [31, 32, 33, 34],
    [41, 42, 43, 44]
])

row_order = np.array([2, 0, 3, 1])
col_order = np.array([3, 1, 0, 2])

print("Оригінал:\n", A)

#1 Переставте рядки відповідно до row_order.
A_rows = A[row_order, :]
print("\nПісля перестановки рядків:\n", A_rows)

#2 Після цього переставте стовпці відповідно до col_order.
A_col = A_rows[:, col_order]
print("\nПісля перестановки стовпців:\n", A_col)

#3 Відновлення оригіналу за допомогою обернених перестановок
inv_col_order = np.argsort(col_order)
inv_row_order = np.argsort(row_order)

A_restored = A_col[:, inv_col_order]
A_restored = A_restored[inv_row_order, :]

print("Відновлена матриця:", A_restored)
print("Чи збігається з оригіналом:", np.array_equal(A_restored, A))
````

## Завдання 12. Трикутні частини матриці

Дано матрицю:
```python
A = np.array([
    [4, 1, 7, 2],
    [6, 3, 5, 8],
    [9, 2, 6, 1],
    [5, 7, 4, 3]
])
```

1. Отримайте верхню трикутну частину разом із головною діагоналлю.
2. Отримайте нижню трикутну частину разом із головною діагоналлю.
3. Отримайте строго верхню та строго нижню частини.
4. Відновіть початкову матрицю з нижньої та строго верхньої частин.

Використайте `np.triu()` та `np.tril()`.

````python
import numpy as np
A = np.array([
    [4, 1, 7, 2],
    [6, 3, 5, 8],
    [9, 2, 6, 1],
    [5, 7, 4, 3]
])

#1 Отримайте верхню трикутну частину разом із головною діагоналлю.
upper = np.triu(A)
print("Верхня трикутна (з діагоналлю):", upper)

#2 Отримайте нижню трикутну частину разом із головною діагоналлю.
lower = np.tril(A)
print("Нижня трикутна (з діагоналлю):", lower)

#3 Отримайте строго верхню та строго нижню частини.
strict_upper = np.triu(A, k=1)
strict_lower = np.tril(A, k=-1)
print("Строго верхня (k=1):", strict_upper)
print("Строго нижня (k=-1):", strict_lower)

#4 Відновіть початкову матрицю з нижньої та строго верхньої частин.
restored = lower + strict_upper
print("Відновлена матриця (lower + strict_upper):", restored)
print("Чи збігається з оригіналом:", np.array_equal(restored, A))
````

````python
A = np.array([
    [2, 1, 0],
    [1, 3, 4],
    [0, 2, 5]
])

B = np.array([
    [1, 3, 5],
    [2, 6, 10],
    [4, 1, 0]
])

#1 Знайдіть ранг кожної матриці.
rank_A = np.linalg.matrix_rank(A)
rank_B = np.linalg.matrix_rank(B)

print("Ранг A:", rank_A)
print("Ранг B:", rank_B)

#2 Визначте, яка з них має повний ранг.
n = A.shape[0]
print()
print("A має повний ранг:", rank_A == n)
print("B має повний ранг:", rank_B == n)
````

## Завдання 19. Власні значення та власні вектори

Дано симетричну матрицю:
```python
A = np.array([
    [6.0, 2.0],
    [2.0, 3.0]
])
```

1. Знайдіть власні значення та власні вектори.
2. Перевірте співвідношення `A @ V = V * λ`.
3. Виведіть результат перевірки через `np.allclose()`.

Використайте `np.linalg.eigh()`.

````python
A = np.array([
    [6.0, 2.0],
    [2.0, 3.0]
])

#1 Знайдіть власні значення та власні вектори.
values, vectors = np.linalg.eigh(A)
print("Власні значення:", values)
print("Власні вектори (стовпці):")
print(vectors)

#2 Перевірте співвідношення A @ V = V * λ.
left_side = A @ vectors
right_side = vectors * values

#3 Виведіть результат перевірки через np.allclose().
print()
print("A @ V =")
print(left_side)
print()
print("V * lambda =")
print(right_side)
print()
print("Рівність виконується:", np.allclose(left_side, right_side))
````

## Завдання 20. Метод найменших квадратів

Для точок
```python
x = np.array([0., 1., 2., 3., 4.])
y = np.array([2.2, 3.8, 6.1, 7.9, 10.2])
```

побудуйте пряму
\[
y \approx a + bx.
\]

1. Сформуйте матрицю ознак зі стовпця одиниць і значень `x`.
2. Знайдіть коефіцієнти `a` і `b` методом найменших квадратів.
3. Обчисліть прогнозні значення `y`.
4. Знайдіть суму квадратів залишків.

Використайте `np.linalg.lstsq()`.

````python
x = np.array([0., 1., 2., 3., 4.])
y = np.array([2.2, 3.8, 6.1, 7.9, 10.2])

#1 Сформуйте матрицю ознак зі стовпця одиниць і значень x.
M = np.column_stack((np.ones_like(x), x))
print("Матриця ознак M:")
print(M)

#2 Знайдіть коефіцієнти a і b методом найменших квадратів.
coeffs, residuals, rank, sv = np.linalg.lstsq(M, y, rcond=None)
a, b = coeffs
print()
print("a =", a)
print("b =", b)

#3 Обчисліть прогнозні значення y.
y_pred = M @ coeffs
print()
print("Прогнозні значення y:", y_pred)
print("Реальні значення y:  ", y)

#4 Знайдіть суму квадратів залишків
sum_res = np.sum((y - y_pred) ** 2)
print()
print("Сума квадратів залишків:", sum_res)
````
