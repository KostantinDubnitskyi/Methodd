```python
max_cor = max(abs(cor_pclass), abs(cor_age), abs(cor_fare))

if max_cor == abs(cor_pclass):
    print('Survived і Pclass мають найвищу кореляцію')
elif max_cor == abs(cor_age):
    print('Survived і Age мають найвищу кореляцію')
else:
    print('Survived і Fare мають найвищу кореляцію')
```

**Частина 2**

1.Завантажуємо дані

```python
url = "https://uk.wikipedia.org/wiki/Населення_України"
html = requests.get(url, headers={"User-Agent": "Mozilla/5.0"}).text

tables = pd.read_html(html)
```

2.Виводимо таблицю

```python
df = tables[12]
df.head()
```

```python
df.columns
```

3.Розмір датасету

```python
df.info()
```

```python
df.dtypes
```

4.Прибираємо [1] та змінюємо назву першого стовпця на Регіон, прибираємо пробіли, а також убираємо Україна з таблиці

```python
df.columns = (df.columns
        .str.replace(r'(?:\s*\[\d+\])+', '', regex=True))
df = df.rename(columns={df.columns[0]: 'Region'})
df
```

```python
df = df.replace("—", np.nan)
```

```python
df = df.drop(df[df.iloc[:, 0] == "Україна"].index)
```

```python
df
```

5. Перетворюем числа на float

```python
for col in df.columns[1:]:
    df[col] = pd.to_numeric(df[col].astype(str).str.replace(',', '.'), errors='coerce')
```

```python
df.dtypes
```

```python
df.isnull().sum()
```

6.Замінив пропущені значення на середнє

```python
for col in df.columns[1:]:
    df[col] = df[col].fillna(df[col].mean())
```

```python
df.isnull().sum()
```

7.Ще раз виводимо датасет

```python
df.tail()
```

8.Дублікати

```python
df.duplicated().sum()
```

9.Статистика

```python
df.describe()
```

10.Різниця між першим і останнім роком для кожного регіону

```python
df['Diff'] = abs(df['1950'] - df['2019'])
max_diff = df['Diff'].max()
```

```python
region = df[df['Diff'] == max_diff]['Region'].values[0]
print(f'{region} має найбільшу різницю між 1950 і 2019: {max_diff}')
```

11.Кореляція між першим і останнім роком

```python
cor_1950_2019 = df['1950'].corr(df['2019'])
print(f'Correlation between 1950 and 2019: {cor_1950_2019}')
```

12.Середні значення за стовпцями

```python
mean_1950 = df['1950'].mean()
mean_1950

mean_2019 = df['2019'].mean()
mean_2019
```

13.Стандарне відхилення

```python
year_cols = list(df.columns[1:-1])
df['std'] = df[year_cols].std(axis=1)
```

14.Регіон, має найбільшу варітивність

```python
df['std'] = df.iloc[:, 1:].std(axis=1)

max_std = df['std'].max()
region = df[df['std'] == max_std]['Region'].values[0]
print(f'{region} має найвищу варіативність: {max_std}')
```

15.Регіони з найвищим та найнижчим показником у кожному році

```python
max_1950 = df['1950'].max(); min_1950 = df['1950'].min()
max_2019 = df['2019'].max(); min_2019 = df['2019'].min()

print(f'Max 1950: {max_1950}, Min 1950: {min_1950}')
print(f'Max 2019: {max_2019}, Min 2019: {min_2019}')
```

ГРАФІК 1: гістограма розподілу за останній рік

```python
plt.hist(df['2019'], bins=10)
plt.title('2019')
plt.xlabel('Коефіцієнт народжуваності')
plt.ylabel('Кількість регіонів')
plt.show()
```

ГРАФІК 2: динаміка топ-10 регіонів за останнім роком

```python
df_top10 = df.nlargest(10, '2019')
plt.bar(df_top10['Region'], df_top10['2019'])
plt.title('Топ-10 регіонів за народжуваністю, 2019')
plt.xticks(rotation=90)
plt.show()
```

ГРАФІК 3: топ-10 регіонів за зміною між першим і останнім роком

```python
df_top10_diff = df.nlargest(10, 'Diff')
plt.bar(df_top10_diff['Region'], df_top10_diff['Diff'])
plt.title('Зміна народжуваності 1950 → 2019 (топ-10)')
plt.xticks(rotation=90)
plt.show()
```
