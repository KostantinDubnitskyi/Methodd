## 19. Підсумкові тренувальні завдання

1. Створіть матрицю `X` форми `(5, 4)` із випадкових чисел.
2. Обчисліть середнє та стандартне відхилення кожної ознаки.
3. Стандартизуйте матрицю через broadcasting.
4. Виберіть рядки, у яких значення першої ознаки додатне.
5. Додайте до матриці новий стовпець одиниць.
6. Створіть вектор ваг і обчисліть прогноз `X @ w`.
7. Для заданого `y_true` обчисліть MSE.
8. Перевірте, чи немає в масивах `NaN` або `inf`.


```python
# 1. Матриця X форми (5, 4) із випадкових чисел
rng = np.random.default_rng(seed=7)
X = rng.normal(loc=50, scale=10, size=(5, 4))

# 2. Середнє та стандартне відхилення кожної ознаки
P = X.mean(axis=0)
S = X.std(axis=0)

print("P:", P)
print("S:", S)

# 3. Стандартизація через broadcasting
X_std = (X - P) / S

print("X_std =", X_std)

# 4. Рядки, у яких значення першої ознаки додатне
mask = X[:, 0] > 0
X_positive = X[mask]

print("Рядки з додатною першою ознакою:", X_positive)

# 5. Додаємо стовпець одиниць
X_d = np.column_stack([
    np.ones(X_std.shape[0]),
    X_std
])

print("X_d =", X_d)

# 6. Вектор ваг і прогноз
w = np.array([1.0, 0.5, -0.3, 0.8, 0.2])
y_pred = X_d @ w

print("y_pred =", y_pred)

# 7. MSE для заданого y_true
y_true = np.array([48., 52., 47., 55., 50.])
mse = np.mean((y_true - y_pred) ** 2)

print("MSE =", mse)

# 8. Перевірка на NaN та inf
print("Є NaN:", np.isnan(X).any())
print("Є inf:", np.isinf(X).any())
```
